```python
# get marks of a student and give a grade based on the marks
# 95 and above-A, 85 and above-B, 75 and above-C, 65 and above-D, below 65-F
marks=int(input('enter your marks:'))
if marks>=95:
    print('your corresponding marks are:', marks, '\nand your grade is:A')
elif marks>=85:
    print('your corresponding marks are:', marks, '\nand your grade is:B')
elif marks>=75:
    print('your corresponding marks are:', marks, '\nand your grade is:C')
elif marks>=65:
    print('your corresponding marks are:', marks, '\nand your grade is:D')
else:
    print('your corresponding marks are:', marks, '\nand your grade is:F')
```
